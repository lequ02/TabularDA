```python
import os
import pandas as pd

# Base directory to start the search
base_directory = r"C:\Users\phuon\DA\src\modeling_thuy\output"

# Subdirectories containing the CSV files (datasets)
datasets = ['mnist12', 'mnist28','covertype','adult','census','credit']

# Columns to analyze
columns_to_analyze = [
    'test_f1_macro', 'test_f1_micro','test_f1_weighted'
]

# Keywords to classify files
train_augment_types = {
    'synthetic': ['ctgan', 'gaussian', 'categorical', 'pca_gmm', 'tvae', 'xgb', 'rf'],
    'original': ['original']
}

# DataFrame to collect results
results_list = []

# Function to classify train & augment option from filename
def classify_train_augment(filename):
    filename_lower = filename.lower()
    for category, keywords in train_augment_types.items():
        for keyword in keywords:
            if keyword in filename_lower:
                return category.capitalize(), keyword  # e.g., ('Mix', 'mix_ctgan')
    return "Unknown", "Unknown"

# Function to process a file
def process_file(file_path, dataset):
    try:
        # Read the CSV file
        df = pd.read_csv(file_path)
        
        # Extract filename and train & augment classification
        original_filename = os.path.basename(file_path)
        train_augment_category, augment_type = classify_train_augment(original_filename)
        
        # Dictionary to store results
        file_results = {'Dataset': dataset, 'Train_Augment_Option': train_augment_category, 'Augment_Type': augment_type}
        
        print("file results: ", file_results)


        # Extract metrics
        for column in columns_to_analyze:
            if column in df.columns:
                file_results[f'{column}_max'] = df[column].max()

                print(file_results[f'{column}_max'])
                file_results[f'{column}_end'] = df[column].iloc[-1]
            # else:
            #     file_results[f'{column}_max'] = None
            #     file_results[f'{column}_end'] = None
                
        
        # print("file results: ", file_results)

        # Append results
        results_list.append(file_results)

        print(results_list)

    except Exception as e:
        print(f"Error processing file {file_path}: {e}")

# Process each dataset folder
for dataset in datasets:
    print("processing dataset: ", dataset)
    subdir_path = os.path.join(base_directory, dataset, 'acc')  # Path to 'acc' folder
    
    if os.path.exists(subdir_path):
        for root, _, files in os.walk(subdir_path):
            for file in files:
                if file.endswith('.csv'):  # Process CSV files only
                    file_path = os.path.join(root, file)
                    print("processing file: ", file_path)
                    process_file(file_path, dataset)
    else:
        print(f"Directory not found: {subdir_path}")

# Convert results to DataFrame
results_df = pd.DataFrame(results_list)
print(results_df)

# Create the pivot table using the metrics columns
final_df = results_df.pivot_table(
    index=['Augment_Type'], 
    columns=['Dataset'],
    values=['test_f1_macro_max', 'test_f1_micro_max', 'test_f1_weighted_max']
)

# Rename the metric columns to more readable names
rename_dict = {
    'test_f1_macro_max': 'Macro',
    'test_f1_micro_max': 'Micro',
    'test_f1_weighted_max': 'Weighted'
}
# Replace level 0 names with their new names
# Instead of using set_levels, build a new MultiIndex with renamed levels
final_df.columns = pd.MultiIndex.from_tuples([
    (rename_dict.get(metric, metric), ds) for metric, ds in final_df.columns
])


# # Swap levels so that the outer level is the metric and inner level is the dataset,
# # then sort by the dataset names
# final_df = final_df.swaplevel(0, 1).sort_index(axis=1, level=0)

# Build the desired column order based on your datasets list
# The order will be: ('Macro', dataset) then ('Micro', dataset) for each dataset.
desired_order = [(metric, ds) for metric in ['Macro', 'Micro','Weighted'] for ds in datasets]

# Reindex the columns to match the desired order
final_df = final_df.reindex(columns=pd.MultiIndex.from_tuples(desired_order))
final_df = final_df.reset_index()

print(final_df)

```

```python
import os
import pandas as pd
import sys

# Add parent directory to the system path to access constants.py
sys.path.insert(0, os.path.abspath(os.path.join(os.getcwd(), '..')))
from constants import IN_DATA_PATHS

# Base directory to start the search
base_directory = r"C:\Users\phuon\DA\output\output"

# Subdirectories containing the CSV files (datasets)
datasets = ['mnist12','mnist28','adult','census','covertype']

# Columns to analyze
columns_to_analyze = [
    'test_f1_macro', 'test_f1_micro'
]

# DataFrame to collect results
results_list = []

# Function to classify dataset and augment type from the filename using constants
def classify_from_filename(filename):
    # Extract dataset and augmentation type from the filename
    filename_lower = filename.lower()
    
    dataset = "Unknown"
    augment_type = "Unknown"
    
    # Loop through the IN_DATA_PATHS to find the matching dataset and augment type
    for key, value in IN_DATA_PATHS.items():
        #print(IN_DATA_PATHS.items())
        # Check if dataset name is in the filename
        print(key)
        print(value)
        if value.get('dataset') and value.get('dataset').lower() in filename_lower:
            dataset = value['dataset']
        
        # Check if augment type is in the filename
        for aug_key in value.get('synthetic', {}):
            if aug_key.lower() in filename_lower:
                augment_type = aug_key
                break  # Stop after the first match

    return dataset, augment_type

# Function to process a file
def process_file(file_path, dataset):
    try:
        # Read the CSV file
        df = pd.read_csv(file_path)
        
        # Check for the columns to analyze in the file
        if not all(column in df.columns for column in columns_to_analyze):
            raise ValueError(f"The CSV file is missing one or more of the required columns: {columns_to_analyze}")
        
        # Extract the dataset name and augmentation type from the filename
        filename = os.path.basename(file_path)
        dataset, augment_type = classify_from_filename(filename)
        
        # Dictionary to store results
        file_results = {'Dataset': dataset, 'Augment_Type': augment_type}

        # Extract metrics and compute max and end values for each metric
        for column in columns_to_analyze:
            if column in df.columns:
                file_results[f'{column}_max'] = df[column].max()
                file_results[f'{column}_end'] = df[column].iloc[-1]
            else:
                raise ValueError(f"Column {column} not found in file {file_path}")
        
        # Append results to the list
        results_list.append(file_results)

    except Exception as e:
        print(f"Error processing file {file_path}: {e}")

# Process each dataset folder
for dataset in datasets:
    subdir_path = os.path.join(base_directory, dataset, 'acc')  # Path to 'acc' folder
    
    if os.path.exists(subdir_path):
        for root, _, files in os.walk(subdir_path):
            for file in files:
                if file.endswith('.csv'):  # Process CSV files only
                    file_path = os.path.join(root, file)
                    print(f"Processing file: {file_path}")
                    process_file(file_path, dataset)
    else:
        print(f"Directory not found: {subdir_path}")

# Convert results to DataFrame
results_df = pd.DataFrame(results_list)
print("Results DataFrame:")
print(results_df)

# Create the pivot table using the metrics columns
final_df = results_df.pivot_table(
    index=['Augment_Type'], 
    columns=['Dataset'],
    values=['test_f1_macro_max', 'test_f1_micro_max']
)

# Rename the metric columns to more readable names
rename_dict = {
    'test_f1_macro_max': 'Macro',
    'test_f1_micro_max': 'Micro'
}
final_df.columns = pd.MultiIndex.from_tuples([  # Reformat column names
    (rename_dict.get(metric, metric), ds) for metric, ds in final_df.columns
])

# Build the desired column order based on your datasets list
desired_order = [(metric, ds) for metric in ['Macro', 'Micro'] for ds in datasets]

# Reindex the columns to match the desired order
final_df = final_df.reindex(columns=pd.MultiIndex.from_tuples(desired_order))
final_df = final_df.reset_index()

# Display the final results DataFrame
print("Final Results DataFrame:")
print(final_df)

# Save results to CSV
final_df.to_csv('results_mar_12.csv', index=False)
```

```python
import os
import re
import pandas as pd

def extract_info(file_path):
    """
    Extract dataset_name, train_option, and augment_option from a file name.
    
    Case 1: File name includes an augment option.
      Expected format: 
      DNN_<dataset_name>_train_<train_option>_augment_<augment_option>_mix_ratio...
    
    Case 2: File name does NOT include an augment option.
      Expected format: 
      DNN_<dataset_name>_train_<train_option>_test_...
      In this case, augment_option will be set to "none".
    """
    file_name = os.path.basename(file_path)
    
    # First try pattern with augment option
    pattern_with_augment = r'^DNN_(?P<dataset_name>.+?)_train_(?P<train_option>.+?)_augment_(?P<augment_option>.+?)_mix_ratio'
    match = re.search(pattern_with_augment, file_name)
    if match:
        return match.group('dataset_name'), match.group('train_option'), match.group('augment_option')
    
    # Fallback: pattern without augment option; assume file has '_train_' and then '_test_'
    pattern_without_augment = r'^DNN_(?P<dataset_name>.+?)_train_(?P<train_option>.+?)_test_'
    match = re.search(pattern_without_augment, file_name)
    if match:
        # In this case, augment option is not provided, so we return "none"
        return match.group('dataset_name'), match.group('train_option'), "none"
    
    print(f"File name format did not match expected patterns for file: {file_name}")
    return None, None, None

def process_csv_file(file_path):
    try:
        df = pd.read_csv(file_path)
    except Exception as e:
        print(f"Error reading {file_path}: {e}")
        return None

    results = {}
    for col in ['test_f1_macro', 'test_f1_micro','test_f1_binary']:
        if col in df.columns:
            max_value = df[col].max()
            end_value = df[col].iloc[-1]
            results[col] = {'max': max_value, 'end': end_value}
        else:
            print(f"Column '{col}' not found in {file_path}.")
    return results

def process_directory(base_dir):
    all_results = {}
    for root, dirs, files in os.walk(base_dir):
        for file in files:
            if file.endswith(".csv"):
                file_path = os.path.join(root, file)
                print(f"Processing {file_path}...")
                dataset_name, train_option, augment_option = extract_info(file_path)
                print(f"Extracted info - Dataset: {dataset_name}, Train: {train_option}, Augment: {augment_option}")
                result = process_csv_file(file_path)
                if result:
                    all_results[file_path] = {'info': (dataset_name, train_option, augment_option),
                                               'metrics': result}
    return all_results

def build_dataframe(results):
    # Convert the results dictionary into a list of rows for the DataFrame
    data = []
    for file_path, details in results.items():
        dataset_name, train_option, augment_option = details['info']
        metrics = details['metrics']
        row = {
            'file_path': file_path,
            'dataset_name': dataset_name,
            'train_option': train_option,
            'augment_option': augment_option,
            'test_f1_macro_max': metrics.get('test_f1_macro', {}).get('max', None),
            'test_f1_macro_end': metrics.get('test_f1_macro', {}).get('end', None),
            'test_f1_micro_max': metrics.get('test_f1_micro', {}).get('max', None),
            'test_f1_micro_end': metrics.get('test_f1_micro', {}).get('end', None),
            'test_f1_binary_max': metrics.get('test_f1_binary', {}).get('max', None),
            'test_f1_binary_end': metrics.get('test_f1_binary', {}).get('end', None)
        }
        data.append(row)
    
    df = pd.DataFrame(data)
    return df

def group_results(df):
    # Create a pivot table grouping by train_option (rows) and using augment_option and dataset_name as columns.
    pivot = df.pivot_table(
        index='train_option',
        columns=['augment_option', 'dataset_name'],
        values=['test_f1_macro_max', 'test_f1_macro_end', 'test_f1_micro_max', 'test_f1_micro_end','test_f1_binary_max', 'test_f1_binary_end'], 
        aggfunc='mean'
    )
    return pivot

    
```

```python
if __name__ == "__main__":
    base_directory = r"C:\Users\phuon\DA\output\output"  # update this path as needed
    results = process_directory(base_directory)
    
    # Build a DataFrame from the results
    df = build_dataframe(results)
    print("\nCombined DataFrame:")
    print(df)
    
    # Group the results: columns will be a MultiIndex with augment type and dataset name.
    # def group_results(df:any) -> pd.DataFrame:
        # Create a pivot table grouping by train_option (rows) and using augment_option and dataset_name as columns.
    grouped_df = group_results(df)
    print("\nGrouped DataFrame (pivot table):")
    grouped_df
```
