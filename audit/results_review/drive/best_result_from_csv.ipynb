```python
import os
import pandas as pd

# Base directory to start the search
base_directory = r"C:\Users\phuon\DA\src\modeling_thuy\output"

# Subdirectories containing the CSV files
subdirectories = ['adult', 'census', 'intrusion', 'mnist12', 'mnist28', 'news']

# Columns to analyze
columns_to_analyze = ['test_accuracy', 'test_f1_macro', 'test_f1_micro', 'test_f1_weighted', 'f1_binary', 'test_loss']

# Keywords to look for in the filename
keywords = ['DNN', 'train', 'augment', 'test']

# DataFrame to collect all results
results_list = []  # Temporary list to hold results

# Function to extract meaningful part of the filename
def extract_keyword_info(filename, keywords):
    for keyword in keywords:
        if keyword in filename:
            # Extract part of the filename after the keyword
            start_index = filename.find(keyword) + len(keyword)
            extracted_part = filename[start_index:].split('_')[0]  # Stop at the next underscore
            return extracted_part
    return "NoKeyword"  # Default if no keyword matches

# Function to process a file and store results in the results list
def process_file(file_path):
    try:
        # Read the CSV file
        df = pd.read_csv(file_path)

        # Extract filename and meaningful part
        original_filename = os.path.basename(file_path)
        extracted_info = extract_keyword_info(original_filename, keywords)

        # Process specified columns and collect the results
        for column in columns_to_analyze:
            if column in df.columns:
                # Find max value and end value
                max_value = df[column].max()
                last_value = df[column].iloc[-1]

                # Append result for this column
                results_list.append({
                    'Filename': original_filename,
                    'Keyword_Info': extracted_info,
                    'Column': column,
                    'Max_Value': max_value,
                    'End_Value': last_value
                })
            else:
                # Append None for missing columns
                results_list.append({
                    'Filename': original_filename,
                    'Keyword_Info': extracted_info,
                    'Column': column,
                    'Max_Value': None,
                    'End_Value': None
                })
    except Exception as e:
        print(f"Error processing file {file_path}: {e}")

# Traverse through each subdirectory and process CSV files
for subdir in subdirectories:
    subdir_path = os.path.join(base_directory, subdir, 'acc')  # Full path to the acc folder

    if os.path.exists(subdir_path):
        for root, dirs, files in os.walk(subdir_path):
            for file in files:
                if file.endswith('.csv'):  # Only process CSV files
                    file_path = os.path.join(root, file)
                    process_file(file_path)
    else:
        print(f"Directory not found: {subdir_path}")

# After processing, create the DataFrame and save the results to a CSV file
results_df = pd.DataFrame(results_list)

# Pivot the DataFrame to structure it by Dataset and Train Option with subcategories
df_pivot = results_df.pivot_table(
    index=['Keyword_Info'],
    columns='Filename',
    values=['Max_Value', 'End_Value'],
    aggfunc='first'
)

# Flatten the MultiIndex in the columns for better readability
df_pivot.columns = [f'{col[0]}_{col[1]}' for col in df_pivot.columns]

# Reset the index to make Keyword_Info part of the columns
df_pivot.reset_index(inplace=True)

# Display the reshaped DataFrame
print(df_pivot)

# Optionally save the result to CSV
df_pivot.to_csv('final_results.csv', index=False)
```
